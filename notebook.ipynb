```python
import pandas as pd
import os

df_name = "results.csv"
df = None

if os.path.exists(df_name):
    df = pd.read_csv(df_name)
else:
    df = pd.DataFrame(columns=["filename", "function_name", "n", "time", "memory", "storage", "marshaling", "no_cache", "hash", "intra_cache", "inter_cache"])
    df.to_csv(df_name, index=False)

df
```

```python
import matplotlib.pyplot as plt

# tempos cenario 1
cenario1_2d_mp_time = df.loc[(df["memory"] == "2d-mp") & (df["inter_cache"] == False), "time"].to_list()
cenario1_2d_ad_time = df.loc[(df["memory"] == "2d-ad") & (df["inter_cache"] == False), "time"].to_list()
x_axis = list(range(len(cenario1_2d_mp_time)))

# plot - cenario 1 - aquele que não tem inter_caching: todas as execuções começam sem o diretorio .intpy

plt.title("Fibonacci time - version comparisons - cenario 1")
plt.plot(x_axis, cenario1_2d_mp_time, color="blue", label="2d-mp")
plt.plot(x_axis, cenario1_2d_ad_time, color="red", label="2d-ad")

plt.xlabel("input (n)")
plt.ylabel("time (seconds)")

plt.legend()
plt.plot()
```

```python
# tempos cenario 2
cenario2_2d_mp_time = df.loc[(df["memory"] == "2d-mp") & (df["inter_cache"] == True), "time"].to_list()
cenario2_2d_ad_time = df.loc[(df["memory"] == "2d-ad") & (df["inter_cache"] == True), "time"].to_list()
x_axis = list(range(len(cenario1_2d_mp_time)))

# # plot cenario 2 - aquele que tem inter_caching: todas as execuções começam com o diretorio .intpy

plt.title("Fibonacci time - version comparisons - cenario 2")
plt.plot(x_axis, cenario2_2d_mp_time, color="blue", label="2d-mp")
plt.plot(x_axis, cenario2_2d_ad_time, color="red", label="2d-ad")

plt.xlabel("input (n)")
plt.ylabel("time (seconds)")

plt.legend()
plt.plot()
```
